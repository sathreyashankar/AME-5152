Problem1: A fair coin is tossed 100 times.Calculate:
- Number of heads
- Number of tails
- Experimental probability of Head
- Experimental probability of Tail

```python
import numpy as np

times_coin_tossed = 100
tosses = np.random.choice(["Heads", "Tails"], times_coin_tossed)

no_of_heads = np.sum(tosses == "Heads")
no_of_tails = np.sum(tosses == "Tails")

prob_heads = no_of_heads / times_coin_tossed
prob_tails = no_of_tails / times_coin_tossed

print('Number of Heads: ', no_of_heads)
print('Number of Tails: ', no_of_tails)
print('Experimental Probability of Head: ', prob_heads)
print('Experimental Probability of Tail: ', prob_tails)
```

Problem 3: Rolling a die: 
> A fair die is rolled 6,000 times.
Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

```python
import numpy as np

n = 6000

outcomes = np.random.randint(1, 7, size=n)

for outcome in range(1, 7):
    count = np.sum(outcomes == outcome)
    probability = count / n

    print(
        "Outcome: ", outcome,
        "==> Count: ", count,
        "==> Experimental Probability: ", probability
    )
```

Problem 4: Addition rule for disjoint events
> A die is rolled 10,000 times.
> Define: 
> - A=getting 1, 
> - B= getting 6, 
> - check the events are joint or disjoint
>> Verify: P(A∪B)=P(A)+P(B)

```python
n = 10000

outcomes = np.random.randint(1, 7, size=n)

A = outcomes == 1
B = outcomes == 6

prob_A = np.mean(A)
prob_B = np.mean(B)

prob_A_union_B = np.mean(A|B)

print('P(A) = ', prob_A)
print('P(B) = ', prob_B)
print('P(A U B) = ', prob_A_union_B)
print('P(A) + P(B) = ', prob_A + prob_B)
```

Problem 5: Addition rule for overlapping events:
> For one die roll, define: 
> - A=getting an even number A={2,4,6} and 
> - B=getting a number greater than 3 B={4,5,6}
> - Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

```python
n = 10000

outcomes = np.random.randint(1, 7, size=n)

A = np.isin(outcomes, [2, 4, 6])
B = np.isin(outcomes, [4, 5, 6])

prob_A = np.mean(A)
prob_B = np.mean(B)

prob_A_intersection_B = np.mean(A & B)
prob_A_union_B = np.mean(A | B)

print('P(A) = ', prob_A)
print('P(B) = ', prob_B)

print('P(A U B) = ', prob_A_union_B)
print('P(A | B) = ', prob_A_intersection_B)

print('P(A) + P(B) - P(A n B) = ', prob_A + prob_B - prob_A_intersection_B)
```

Problem 6: Multiplication rule for independent events: 

> A coin is tossed and a die is rolled 10,000 times.
> Define:
>> - obtaining Head
>> - obtaining an even number
>> - Verify: P(A∩B)=P(A)P(B)

```python
n = 10000

coin = np.random.choice(["H", "T"], size=n)
die = np.random.randint(1, 7, size=n)

A = coin == "H"
B = die % 2 == 0

prob_A = np.mean(A)
prob_B = np.mean(B)
prob_intersection = np.mean(A & B)

print("P(A) = ", prob_A)
print("P(B) = ", prob_B)
print("P(A ∩ B) = ", prob_intersection)

print("P(A) × P(B) = ", prob_A * prob_B)
```
