```python
import numpy as numpy
import pandas as pd
import matplotlib.pyplot as plt
```

```python
probability = 1

no_of_people = int(input('Enter no of people: '))
data = []

for i in range(1, no_of_people+1):
    for j in range(1, i+1):
        probability = probability*((365-j+1)/365)

    data.append({
        'No Of People': i,
        'Probability': probability
    })

probability = 1
df = pd.DataFrame(data)
print(df)
```

```python
plt.plot(df['No Of People'], df['Probability'])
plt.xlabel('No of People')
plt.ylabel('Probability')
plt.title('Birthday Problem')
plt.grid()
plt.show()
```
